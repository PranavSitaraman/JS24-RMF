```python
import polars as pl
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
date_offset = 1690
is_score_dates = 5
pl_all = pl.scan_parquet(
    "./input/jane-street-real-time-market-data-forecasting/train.parquet"
).filter(pl.col("date_id") >= date_offset - 1)
pl_all = pl_all.drop("partition_id")
```

```python
syn_test = pl_all.with_columns(pl.col("date_id") - date_offset)
syn_test = syn_test.with_columns(
    pl.when(pl.col("date_id") < is_score_dates - 1)
    .then(pl.lit(False))
    .otherwise(pl.lit(True))
    .alias("is_scored")
)
syn_test = syn_test.filter(pl.col("date_id") >= 0).filter(
    pl.col("date_id") <= 8
)
syn_test = syn_test.select(
    ["date_id", "time_id", "symbol_id", "weight", "is_scored"]
    + [f"feature_{x:02}" for x in range(79)]
).with_row_index(name="row_id", offset=0)
syn_test.collect().write_parquet("./input/tuning/synthetic_test.parquet")
syn_test.collect()
```

```python
syn_lag = pl_all.select(
    ["date_id", "time_id", "symbol_id"] + [f"responder_{x}" for x in range(9)]
).with_columns(pl.col("date_id") - date_offset + 1)
syn_lag = syn_lag.rename(
    {f"responder_{x}": f"responder_{x}_lag_1" for x in range(9)}
)
syn_lag = syn_lag.filter(pl.col("date_id") >= 0).filter(pl.col("date_id") <= 8)
syn_lag.collect().write_parquet(
    f"./input/tuning/synthetic_lag.parquet",
)
syn_lag.collect()
```

```python
syn_valid = pl_all.select(
    ["date_id", "time_id", "symbol_id", "weight", "responder_6"]
).with_columns(pl.col("date_id") - date_offset)
syn_valid = syn_valid.filter(pl.col("date_id") >= 0).filter(
    pl.col("date_id") <= 8
)
syn_valid.collect().write_parquet("./input/tuning/synthetic_valid.parquet")
syn_valid.collect()
```

```python
ans = pl.read_parquet("./input/tuning/ans_0.parquet")
for i in range(1, 9):
    ans = pl.concat([ans, pl.read_parquet(f"./input/tuning/ans_{i}.parquet")])
ans
```

```python
responder_6_valid = syn_valid.collect()["responder_6"].to_numpy()
sample_weights = syn_valid.collect()["weight"].to_numpy()
X = ans.select(["nn_xgb", "ridge", "tabm"]).to_numpy()

regressor = LinearRegression()
regressor.fit(X, responder_6_valid, sample_weight=sample_weights)

weights = np.append(regressor.coef_, regressor.intercept_)
print("Ensemble weights:", weights)
```
